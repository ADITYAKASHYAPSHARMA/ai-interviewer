# ⚡ Aegis AI Agent — Interactive Interviewer Tutorial

This notebook demonstrates how the **Aegis AI Agent** works behind the scenes using **LangChain**, multi-provider LLMs (**Groq**, **Google Gemini**, **OpenAI**), dynamic conversation state management, and the Hiring Committee evaluation chain.

```python
# Step 1: Import core dependencies and load environment variables
import os
from dotenv import load_dotenv

load_dotenv()

print("Environment status:")
print("• GROQ_API_KEY:", "✅ Present" if os.getenv("GROQ_API_KEY") else "❌ Missing")
print("• GOOGLE_API_KEY:", "✅ Present" if os.getenv("GOOGLE_API_KEY") else "❌ Missing")
print("• OPENAI_API_KEY:", "✅ Present" if (os.getenv("OPENAI_API_KEY") or os.getenv("OPENNAI_API_KEY")) else "❌ Missing")
```

```python
# Step 2: Initialize the Interview Engine
from aiagent.agent import agent_engine, AVAILABLE_ROLES, AVAILABLE_PERSONAS

print("Active Engine Provider:", agent_engine.active_provider)
print(f"Loaded {len(AVAILABLE_ROLES)} preset roles and {len(AVAILABLE_PERSONAS)} interviewer personas.")
```

```python
# Step 3: Create an Interview Session
session = agent_engine.create_session(
    candidate_name="Aditya Sharma",
    role="AI / ML & LLM Engineer",
    level="Senior",
    persona="FAANG Bar Raiser",
    total_questions=3
)

print(f"Session Created! ID: {session.id}")
print(f"Candidate: {session.candidate_name} | Role: {session.level} {session.role}")
print(f"Interviewer Persona: {session.persona}")
```

```python
# Step 4: Start the Interview and get Question 1
start_result = agent_engine.start_interview(session.id)

print("🤖 INTERVIEWER:")
print("-" * 60)
print(start_result["message"])
print("-" * 60)
```

```python
# Step 5: Answer Question 1 and receive Question 2
candidate_answer_q1 = (
    "In our production RAG system, we implemented hybrid dense-sparse retrieval using "
    "Qdrant for vector embeddings and BM25 for keyword search, merged via Reciprocal Rank Fusion (RRF). "
    "To minimize hallucination, we integrated cross-encoder rerankers and contextual self-consistency evaluation."
)

print(f"👤 CANDIDATE:\n{candidate_answer_q1}\n")

response_q1 = agent_engine.process_response(session.id, candidate_answer_q1)

print("🤖 INTERVIEWER:")
print("-" * 60)
print(response_q1["message"])
print("-" * 60)
```

```python
# Step 6: Answer Question 2
candidate_answer_q2 = (
    "When fine-tuning open-source LLMs under constrained compute, I utilize QLoRA with 4-bit NormalFloat quantization "
    "and rank-16 adapters. We track loss and perplexity in Weights & Biases and validate against an unseen golden test set."
)

print(f"👤 CANDIDATE:\n{candidate_answer_q2}\n")

response_q2 = agent_engine.process_response(session.id, candidate_answer_q2)

print("🤖 INTERVIEWER:")
print("-" * 60)
print(response_q2["message"])
print("-" * 60)
```

```python
# Step 7: Answer Final Question 3 & Wrap up
candidate_answer_q3 = (
    "For latency-critical LLM inference, we deploy vLLM with PagedAttention, KV-cache quantization, and speculative decoding, "
    "achieving a 3.4x throughput boost and keeping p95 TTFT (Time To First Token) under 180ms."
)

print(f"👤 CANDIDATE:\n{candidate_answer_q3}\n")

response_q3 = agent_engine.process_response(session.id, candidate_answer_q3)

print("🤖 INTERVIEWER:")
print("-" * 60)
print(response_q3["message"])
print("-" * 60)
```

```python
# Step 8: Generate Hiring Committee Evaluation Scorecard
evaluation = agent_engine.evaluate_interview(session.id)

import json
print("=" * 60)
print(f"🏆 OFFICIAL SCORECARD: {evaluation.get('verdict', 'N/A').upper()} ({evaluation.get('overall_score', 0)}/100)")
print("=" * 60)
print("Executive Summary:", evaluation.get("summary"))
print("\nSkill Pillars:", json.dumps(evaluation.get("metrics"), indent=2))
print("\nStrengths:")
for s in evaluation.get("strengths", []):
    print(f"  • {s}")
print("\nAreas for Improvement:")
for imp in evaluation.get("areas_for_improvement", []):
    print(f"  • {imp}")
```
